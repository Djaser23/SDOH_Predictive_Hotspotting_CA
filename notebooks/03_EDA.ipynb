### Rent burden rate: choice of denominator

The rent burden rate is the share of renter households paying 30% or more of income on rent. The numerator is the four brackets at 30% and above (`rent_burden_30_35`, `rent_burden_35_40`, `rent_burden_40_50`, `rent_burden_50plus`).

The denominator is the sum of the nine computed brackets, not `renter_hh_total`. `renter_hh_total` (B25070_001E) is the full renter count and matches `renter_occupied_units` in all 9,070 tracts. In the typical tract it is about 21 households higher than the nine brackets combined (median gap 21, maximum 541). The Census's documentation says units paying no rent and households reporting no income or a net loss fall into a separate "Not computed" category ([Social Explorer copy of the B25070 metadata, 2012–2016 version](https://www.socialexplorer.com/data/ACS2016_5yr/metadata/?ds=ACS16_5yr&table=B25070)). The gap is consistent with that category, but the extraction notebook did not pull B25070_011E, so I have not checked it directly.

Dividing by `renter_hh_total` would count those households as not burdened and lower the rate. Tracts with zero computed households are set to NaN.

```python
# check similarity between renter_hh_total and renter_occupied_units
df[['renter_hh_total','renter_occupied_units']].head()
```

```python
# check if the sum of rent brackets equals renter_hh_total
# check if renter_hh_total has the same data as renter_occupied_units
# inspect rent_not_computed column

brackets = ['rent_burden_lt10', 'rent_burden_10_15', 'rent_burden_15_20',
            'rent_burden_20_25', 'rent_burden_25_30', 'rent_burden_30_35',
            'rent_burden_35_40', 'rent_burden_40_50', 'rent_burden_50plus']
df['rent_not_computed'] = df['renter_hh_total'] - df[brackets].sum(axis=1)

print(df['rent_not_computed'].describe())
print((df['renter_hh_total'] == df['renter_occupied_units']).mean())
df.loc[[1319, 9067], ['renter_hh_total', 'rent_not_computed']]
```

```python
# create rent burden rate
burden_cols = ['rent_burden_30_35', 'rent_burden_35_40', 'rent_burden_40_50', 'rent_burden_50plus']

df['rent_computed'] = df[brackets].sum(axis=1)
df['rent_burden_rate'] = df[burden_cols].sum(axis=1) / df['rent_computed'].where(df['rent_computed'] > 0)
df['rent_burden_rate'].describe()
```

```python
# build no high school diploma rate
# with a numerator being the sum of all columns populations of persons 25+ years old
# with less than a high school diploma and the denominator the total population of 25+ years old

no_hs_dip_cols = ['edu_no_school', 'edu_nursery', 'edu_kindergarten', 'edu_grade_1', 'edu_grade_2',
'edu_grade_3', 'edu_grade_4', 'edu_grade_5', 'edu_grade_6', 'edu_grade_7', 'edu_grade_8',
'edu_grade_9', 'edu_grade_10','edu_grade_11', 'edu_12th_no_diploma']

df['no_hs_dip_rate'] = df[no_hs_dip_cols].sum(axis=1) / df['edu_total_25plus'].where(df['edu_total_25plus'] > 0) 
df['no_hs_dip_rate'].describe()
```

```python
# investigate tracts with high rate of no hs diploma
print((df['edu_total_25plus'] < 30).sum(), 'tracts with fewer than 30 adults aged 25+')

print('Tracts with rates of no hs diploma above 0.7...')
df.loc[df['no_hs_dip_rate'] > .70, ['NAME', 'no_hs_dip_rate', 'rent_burden_rate','edu_total_25plus', 
                                    'total_pop', 'median_age']]
```

### Tracts with no-diploma rate above 0.7

Nine tracts have rates above 70%. Eight are mid-sized (1,000-3,300 adults 25+) with young median ages (22-38), clustered in Monterey, Riverside, and Santa Barbara counties. Their locations appear to be agricultural areas, but I have not verified that. With 1,000 to 3,300 adults each, these rates are unlikely to be small-denominator noise.

The San Francisco tract 9805.01 (151 adults 25+, median age 79.9) has the highest values in the data for diabetes, high blood pressure, coronary heart disease, COPD, and stroke. Its 9800-series number suggests a special-use tract, but I have not verified what it contains. Whether to keep, flag, or drop it is still open.

### Tracts with extreme unemployment rates

Two tracts have unemployment rates above 50%.

- Los Angeles tract 9800.33: 6 of 6 in the labor force unemployed (14 residents). With a denominator that small, one person changes the rate by about 17 points, so this value is noise. It is a 9800-series tract that also appears in the top five for COPD and cognitive disability.
- Imperial County tract 104.01: 866 of 1,526 unemployed (56.7%). The denominator is large, so this is not small-sample noise. The cause is not verified. The margin of error has not been checked.

```python
# build vacancy rate: vacant housing units / total housing units (zero-guarded)
df['vacancy_rate'] = df['housing_units_vacant'] / df['housing_units_total'].where(df['housing_units_total'] > 0)
df['vacancy_rate'].describe()
```

```python
# investigate tracts with high vacancy rate
print((df['vacancy_rate'] > 0.4).sum(), 'tracts with vacancy rate over 40%')
print((df['vacancy_rate'] > 0.7).sum(), 'tracts with vacancy rate over 70%')
df.loc[df['vacancy_rate'] > 0.7,['NAME', 'housing_units_total', 'housing_units_vacant', 'vacancy_rate', 'median_hh_income', 'total_pop']]
```

### Tracts with high vacancy rates

138 tracts have vacancy rates above 40%, and 27 are above 70%. Most of the 27 are in mountain and resort counties: Placer, El Dorado, Nevada, Mono, Calaveras, Tuolumne, Plumas, Alpine, Fresno, and San Bernardino. Their median household incomes run from about $56k to $204k, mostly middle to upper income.

Census counts seasonal and vacation homes as vacant, so a high vacancy rate here may reflect second homes rather than economic strain. Income alone can't separate the two, because second homes and short-term rentals also exist in lower-income rural areas. Table B25004 (vacancy status) would allow a vacancy rate that excludes seasonal units (B25004_006E). How Census classifies full-time short-term rentals has not been checked.

For now, `vacancy_rate` stays as is. Its meaning is ambiguous, which will matter when interpreting its SHAP values.

Two Riverside tracts may not fit the vacation-home pattern: 465.01 (58 housing units but 5,727 residents, which suggests group quarters) and 9401 (190 residents). Neither has been verified.
